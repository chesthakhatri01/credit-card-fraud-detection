```python
import joblib
import pandas as pd
```

```python
model = joblib.load('fraud_detection_model.pkl')
scaler = joblib.load('amount_scaler.pkl')
```

```python

new_transaction = {
    'V1': -1.359807, 'V2': -0.072781, 'V3': 2.536347, 'V4': 1.378155,
    'V5': -0.338321, 'V6': 0.462388, 'V7': 0.239599, 'V8': 0.098698,
    'V9': 0.363787, 'V10': 0.090794, 'V11': -0.551600, 'V12': -0.617801,
    'V13': -0.991390, 'V14': -0.311169, 'V15': 1.468177, 'V16': -0.470401,
    'V17': 0.207971, 'V18': 0.025791, 'V19': 0.403993, 'V20': 0.251412,
    'V21': -0.018307, 'V22': 0.277838, 'V23': -0.110474, 'V24': 0.066928,
    'V25': 0.128539, 'V26': -0.189115, 'V27': 0.133558, 'V28': -0.021053,
    'Amount': 149.62
}

```

```python
df_new = pd.DataFrame([new_transaction])
df_new['Amount_scaled'] = scaler.transform(df_new[['Amount']])
df_new = df_new.drop('Amount', axis=1)

```

```python
prediction = model.predict(df_new)
probability = model.predict_proba(df_new)

if prediction[0] == 1:
    print(f"FRAUD detected! Confidence: {probability[0][1]*100:.2f}%")
else:
    print(f"Genuine transaction. Confidence: {probability[0][0]*100:.2f}%")
```
